## 2. Os verificadores

- **Citações**: toda referência `arquivo.(log|jsonl|csv|txt):N` precisa existir e a linha precisa conter o valor que a frase afirma (você confere com os olhos; o script mostra a linha).
- **Superafirmação**: palavras que afirmam mais do que a evidência, em inglês e português.
- **Termo único**: cada `### Achado` (ou `### Resposta ao quesito`) usa exatamente um termo da escala; "Confirmado" precisa citar pelo menos dois arquivos diferentes.

```python
import re

CITACAO = re.compile(r"([A-Za-z0-9._-]+\.(?:log|jsonl|csv|txt)):(\d+)")
SUPERAFIRMA = re.compile(r"\b(proves?|clearly|obviously|stole|hacker|attacker|prova(?:m)?|claramente|obviamente|"
                         r"roub(?:ou|aram)|atacante|invasor|criminoso|certamente|com certeza)\b", re.I)
TERMOS = ["Confirmado", "Altamente provável", "Provável", "Possível", "Não determinado"]


def checar_citacoes(texto):
    arquivos = {p.name: p for p in WORK.rglob("*") if p.is_file()}
    problemas = 0
    for nome, n in sorted(set(CITACAO.findall(texto)), key=lambda x: (x[0], int(x[1]))):
        p = arquivos.get(nome)
        if p is None:
            print(f"  {nome}:{n}  FILE NOT FOUND"); problemas += 1; continue
        linhas = p.read_text(encoding="utf-8").splitlines()
        if not 0 < int(n) <= len(linhas):
            print(f"  {nome}:{n}  LINE DOES NOT EXIST"); problemas += 1; continue
        print(f"  {nome}:{n}  {linhas[int(n) - 1][:95]}")
    return problemas


def checar_palavras(texto):
    achados = [(i, m.group()) for i, l in enumerate(texto.splitlines(), 1) for m in SUPERAFIRMA.finditer(l)]
    for i, palavra in achados:
        print(f"  linha {i}: '{palavra}'")
    return len(achados)


def checar_termos(texto):
    problemas = 0
    blocos = re.split(r"^### ", texto, flags=re.M)[1:]           # cada achado começa com '### '
    for b in blocos:
        titulo = b.splitlines()[0]
        if not re.match(r"(Achado|Resposta ao quesito)", titulo):
            continue
        # 'Provável' também aparece dentro de 'Altamente provável': conte o mais longo primeiro
        resto = titulo
        usados = []
        for t in TERMOS:
            if re.search(re.escape(t), resto, re.I):
                usados.append(t); resto = re.sub(re.escape(t), "", resto, flags=re.I)
        arquivos = {a for a, _ in CITACAO.findall(b)}
        if len(usados) != 1:
            print(f"  ❌ '{titulo[:60]}': {len(usados)} termos {usados}"); problemas += 1
        elif usados[0] == "Confirmado" and len(arquivos) < 2:
            print(f"  ❌ '{titulo[:60]}': Confirmado com {len(arquivos)} arquivo(s) citado(s)"); problemas += 1
        else:
            print(f"  ✅ '{titulo[:60]}' → {usados[0]} ({len(arquivos)} arquivo(s))")
    return problemas


def verificar(texto, nome="relatório"):
    print(f"=== {nome}\n[citações]"); c = checar_citacoes(texto)
    print("[superafirmação]"); p = checar_palavras(texto)
    print("[termos de confiança]"); t = checar_termos(texto)
    print(f"→ {c} citação(ões) quebrada(s), {p} palavra(s) a revisar, {t} achado(s) mal calibrado(s)\n")
    return c + p + t
```

## 6. 🎮 Quiz: qual termo de confiança?

Checkpoint BR: pelo menos 5 de 6.

```python
# ------------------------------------------------------------
# QUIZ: função genérica de múltipla escolha
# MODO_JOGO = True  -> pergunta e conta acertos
# MODO_JOGO = False -> só mostra gabarito e explicação
# ------------------------------------------------------------
def jogar_quiz(perguntas, modo_jogo=True, minimo=None):
    acertos = 0                                            # placar
    for n, (enunciado, opcoes, certa, porque) in enumerate(perguntas, start=1):
        print(f"\n[{n}/{len(perguntas)}] {enunciado}")    # o enunciado
        for letra, texto in opcoes.items():                # as alternativas a), b), c)
            print(f"   {letra}) {texto}")
        if modo_jogo:
            resposta = ""
            for _ in range(5):                             # até 5 tentativas de digitar uma letra válida
                resposta = input("   Sua resposta: ").strip().lower()
                if resposta in opcoes:
                    break
            ok = (resposta == certa)                       # acertou?
            acertos += ok
            print("   ✅ Acertou!" if ok else f"   ❌ Era {certa}).")
        else:
            print(f"   Gabarito: {certa})")
        print(f"   Por quê: {porque}")
    if modo_jogo:
        print(f"\n🏁 Resultado: {acertos}/{len(perguntas)}")
        if minimo is not None:
            print("Checkpoint BR cumprido. 🥷" if acertos >= minimo
                  else f"Checkpoint BR pede pelo menos {minimo}. Releia o README e tente de novo.")
    return acertos
```
